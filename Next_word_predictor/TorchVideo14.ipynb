```python
import torch
import torch.nn as n
import torch.optim as optim
import numpy as np
from collections import Counter
from torch.utils.data import Dataset,DataLoader
from nltk.tokenize import word_tokenize
import nltk
```

```python
with open("document.text", "r", encoding="utf-8") as f:
    document= f.read()
document
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
#tokenize
tokens=word_tokenize(document.lower())
```

```python
vocab={'<UNK>':0}

for token in Counter(tokens).keys():
  if token not in vocab:
    vocab[token]=len(vocab)
```

```python
len(vocab)
```

```python
input_sentences=[]
```

```python
input_sentences=document.split('\n')
```

```python
def text_to_index(sentence,vocab):
  numerical_sentence=[]

  for token in sentence:
    if token in vocab:
      numerical_sentence.append(vocab[token])
    else:
      numerical_sentence.append(vocab['<UNK>'])
  return numerical_sentence
```

```python
input_numberized_sentences=[]

for sentences in input_sentences:
  input_numberized_sentences.append(text_to_index(word_tokenize(sentences.lower()),vocab))
```

```python
input_numberized_sentences
```

```python
len(input_numberized_sentences)
```

```python
training_sequence=[]
for sentences in input_numberized_sentences:

  for i in range(1,len(sentences)):
    training_sequence.append(sentences[:i+1])
```

```python
len_list=[]
for sequence in training_sequence:
  len_list.append(len(sequence))
```

```python
max(len_list)
```

```python
padded_trianing_sequence=[]
for sequence in training_sequence:
  padded_trianing_sequence.append(([0]*(max(len_list)-len(sequence)))+sequence)
```

```python
padded_trianing_sequence=torch.tensor(padded_trianing_sequence)
x=padded_trianing_sequence[:,:-1]
y=padded_trianing_sequence[:,-1]
```

```python
class CustomDataset(Dataset):
  def __init__(self,x,y):
    self.x=x
    self.y=y

  def __len__(self):
    return self.x.shape[0]

  def __getitem__(self, index):
    return self.x[index],self.y[index]
```

```python
dataset=CustomDataset(x,y)
```

```python
dataset[0][0].shape
```

```python
dataloader=DataLoader(dataset,batch_size=32,shuffle=True)
```

```python
for input , output in dataloader:
  print(input.shape,output.shape)
```

```python
import torch.nn as nn
class LSTModel(nn.Module):
  def __init__(self,vocab_size):
    super().__init__()
    self.embedding=nn.Embedding(vocab_size,100)
    self.lstm=nn.LSTM(100,150,batch_first=True)
    self.fc=nn.Linear(150,vocab_size)
  def forward(self,x):
    embedded=self.embedding(x)
    intermediate_hidden_states,(final_hidded_state,final_cell_state)=self.lstm(embedded)
    output=self.fc(final_hidded_state.squeeze(0))
    return output
```

```python
model=LSTModel(len(vocab))
device=torch.device('cuda' if torch.cuda.is_available() else "cpu")
```

```python
model.to(device)
```

```python
epochs=50
learning_rate=0.001

criterion=nn.CrossEntropyLoss()

optimizer=torch.optim.Adam(model.parameters(),lr=learning_rate)
```

```python
for epoch in range(epochs):
    total_loss=0

    for batch_x,batch_y in dataloader:

      batch_x,batch_y=batch_x.to(device),batch_y.to(device)

      optimizer.zero_grad()

      output=model(batch_x)

      loss=criterion(output,batch_y)

      loss.backward()

      optimizer.step()

      total_loss=total_loss+ loss.item()
    print(f"Epoch: {epoch+1}, Loss: {total_loss:4f}")
```

```python
def prediction(model,vocab,text):

  #tokenize
  tokenized_text=word_tokenize(text.lower())

  #numericalize
  numericalized_text=text_to_index(tokenized_text,vocab)

  #padding

  padded_text=torch.tensor([([0]*(max(len_list)-len(numericalized_text)))+numericalized_text],dtype=torch.long)

  # print(padded_text.shape)
  #send to model
  output=model(padded_text)

  #perdicat index
  value,index=torch.max(output,dim=1)

  return text+" "+list(vocab.keys())[index]
```
